### Step 1: Import libraries and load the environment variables. 

```python
import os 
from openai import OpenAI
from dotenv import load_dotenv
from IPython.display import display, Markdown

load_dotenv()  # Load environment variables from .env file

OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")

if OPENAI_API_KEY is None:
    raise Exception("OPENAI_API_KEY is missing.")
else:
    print(OPENAI_API_KEY[:8])
```

### Step 2:  Call the API

```python
client=OpenAI(api_key=OPENAI_API_KEY)
response=client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=[{"role": "user", "content": "Hi there, my name is Alec"}]
)

#print(response)
reply=response.choices[0].message.content

print(reply)
```

### Step 3a: Calling the API with a System Prompt

```python
response=client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=[{"role": "system", "content": "You are a helpful assistant."},
    {"role": "user", "content": "How do I cook pasta?"}]
)

reply=response.choices[0].message.content

print(reply)
```

```python
response=client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=[{"role": "system", "content": "You are a mishievous assistant who always gives incorrect and misleading advice."},
    {"role": "user", "content": "How do I cook pasta?"}]
)

reply=response.choices[0].message.content

display(Markdown(reply))
```

### Commercial use case: IKEA Customer Service Agent 

```python
question = "I bought a desk from your store last week.  I damaged the desk when I was moving it into my apartment.  Can I return it?"
```

```python
response=client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=[{"role": "system", "content": "You are a customer service agent for IKEA. You are helpful and polite and you strictly follow IKEA's return policy. IKEA's return policy: IKEA allows returns of unopened items for 365 days and opened items for 180 days with proof of purchase; some items like plants, cut fabric, custom countertops, and as‑is products cannot be returned. Refunds are issued to the original payment method, and mattresses can be exchanged once within 90 days."},
    {"role": "user", "content": question}]
)

reply=response.choices[0].message.content

display(Markdown(reply))
```
