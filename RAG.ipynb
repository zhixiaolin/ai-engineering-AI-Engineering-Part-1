### Step 3: Chunk the document

```python
def split_text_into_chunks(text, chunk_size=500, overlap=50):
    """
    Split a long text into overlapping chunks with natural boundary detection.
    
    Args:
        text (str): The input text to split.
        chunk_size (int): Maximum size of each chunk (default 500).
        overlap (int): Number of characters to overlap between chunks (default 50).
    
    Returns:
        list[str]: List of text chunks.
    """
    if not text:
        return []
    if overlap >= chunk_size:
        raise ValueError("overlap must be smaller than chunk_size")
    
    chunks = []
    start = 0
    text_len = len(text)
    
    while start < text_len:
        end = min(start + chunk_size, text_len)
        
        # If we're not at the very end, try to find a natural boundary
        if end < text_len:
            halfway = start + chunk_size // 2
            segment = text[start:end]
            
            # Look for boundaries in priority order, only past halfway point
            boundary_offset = None
            
            # 1. Paragraph break ("\n\n")
            idx = segment.rfind("\n\n")
            if idx != -1 and (start + idx) >= halfway:
                boundary_offset = idx + 2  # include the paragraph break
            
            # 2. Newline ("\n")
            if boundary_offset is None:
                idx = segment.rfind("\n")
                if idx != -1 and (start + idx) >= halfway:
                    boundary_offset = idx + 1
            
            # 3. Sentence end (". ", "! ", "? ")
            if boundary_offset is None:
                best = -1
                for punct in (". ", "! ", "? "):
                    idx = segment.rfind(punct)
                    if idx > best:
                        best = idx
                if best != -1 and (start + best) >= halfway:
                    boundary_offset = best + 1  # keep punctuation, drop space
            
            # 4. Space (" ")
            if boundary_offset is None:
                idx = segment.rfind(" ")
                if idx != -1 and (start + idx) >= halfway:
                    boundary_offset = idx
            
            if boundary_offset is not None:
                end = start + boundary_offset
        
        chunk = text[start:end]
        chunks.append(chunk)
        
        # If we've reached the end of the text, stop
        if end >= text_len:
            break
        
        # Move start forward, leaving `overlap` characters of overlap
        start = end - overlap
        if start < 0:
            start = 0
    
    return chunks
```
