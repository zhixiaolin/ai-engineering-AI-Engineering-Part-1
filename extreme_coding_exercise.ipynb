### Step 4: Use GPT to summarize the replies from three locally deployed LLMs

```python
Topic_Context={
    "reply from GLM 4.6": reply_glm,
    "reply from gemma 4": reply_gemma,
    "reply from Qwen 3": reply_qwen

}
```

```python
Topic_Context={
    "GLM 4.6": reply_glm,
    "gemma 4": reply_gemma,
    "Qwen 3": reply_qwen,
}

system_message = (
    "You are a helpful assistant that summarizes and synthesizes "
    "recommendations from multiple LLMs into a cohesive overview."
)

message="""I am an experienced data scientist. Now I want to expand my horizons to learn about data 
ingestion, data storage, and database maintenance. I plan to use my own computer for hands-on practice. 
Three LLMs have separately provided me with suggestions.  Now I want to summarize their suggestions.
"""

def respond_ai(message):
    # Inject the dynamic context based on the keywords of the messsge
    system_message_enhanced = system_message
    for keyword, context in Topic_Context.items():
            system_message_enhanced += f"\n\n--- suggestion from {keyword} ---\n {context}"

    print(system_message_enhanced)
            
    messages = [{"role": "system", "content": system_message_enhanced},
                {"role": "user", "content": message}]
    
    client = OpenAI(api_key=OPENAI_API_KEY)
    response = client.chat.completions.create(
        model="gpt-4.1-mini",
        messages=messages
    )
    
    return response.choices[0].message.content

reply_summary=respond_ai(message)

display(Markdown(reply_summary))
#gr.ChatInterface(respond_ai).launch(inbrowser=True)
```

```python
from openai import OpenAI
from IPython.display import Markdown, display

# --- Context: the three LLM replies ---
Topic_Context = {
    "GLM 4.6": reply_glm,
    "gemma 4": reply_gemma,
    "Qwen 3":  reply_qwen,
}

# --- System message ---
system_message = (
    "You are a helpful assistant that summarizes and synthesizes "
    "recommendations from multiple LLMs into a cohesive overview."
)

# --- User message (no need to mention keywords) ---
message = """I am an experienced data scientist. Now I want to expand my horizons to learn about data
ingestion, data storage, and database maintenance. I plan to use my own computer for hands-on practice.
Three LLMs have separately provided me with suggestions. Now I want to summarize their suggestions.
"""

def respond_ai(message):
    # Start from the base system message
    system_message_enhanced = system_message

    # Inject EVERY reply as additional context (no keyword matching needed)
    for name, context in Topic_Context.items():
        system_message_enhanced += f"\n\n--- Suggestions from {name} ---\n{context}"

    messages = [
        {"role": "system", "content": system_message_enhanced},
        {"role": "user",   "content": message},
    ]

    client = OpenAI(api_key=OPENAI_API_KEY)
    response = client.chat.completions.create(
        model="gpt-4.1-mini",
        messages=messages,
    )
    return response.choices[0].message.content

reply_summary = respond_ai(message)
display(Markdown(reply_summary))
```
