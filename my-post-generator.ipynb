# Step 1: Import libraries and API keys

```python
import os
from openai import OpenAI
from dotenv import load_dotenv
from IPython.display import display, Markdown

load_dotenv()

OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")

if OPENAI_API_KEY is None:
    raise Exception("OPENAI_API_KEY environment variable is not set.")
```

# Step 2: Search the web

```python
client = OpenAI(api_key=OPENAI_API_KEY)

completion = client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=[
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": "Select three top news stories about China published on reliable sources in past week, and summarize each."}
    ],
    temperature=0.7,
    max_tokens=500
)
```

```python
search_results=completion.choices[0].message.content
display(Markdown(search_results))
```

```python
client = OpenAI(api_key=OPENAI_API_KEY)

completion = client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=[
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": "Select 5 remote job opportunities for senior data scientist positions in the U.S., published on reliable sources in past week, and summarize each."}
    ],
    temperature=0.7,
    max_tokens=500
)

search_results=completion.choices[0].message.content
display(Markdown(search_results))
```

### Step 3: Summarize and write social media post

```python
system_message = """
You are a helpful assistant who creates amazing linkedin posts about AI based on text that you are provided with.  
Make the posts engaging, informative, and professional.  Use a friendly and approachable tone.  Include relevant hashtags at the 
end of the post.
"""

prompt = f"""
Create a linkedin post about the following text.  Make the post engaging, informative, and professional.  Use a friendly and approachable tone.  Include relevant hashtags at the end of the post.
{search_results}

"""


completion = client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=[
        {"role": "system", "content": system_message},
        {"role": "user", "content": prompt.format(search_results=search_results)}
    ],
    temperature=0.7,
    max_tokens=500
    )

chat_response=completion.choices[0].message.content
display(Markdown(chat_response))
```

```python
system_message = """
You are a helpful assistant who creates posts about news stories that you are provided with.  
Make the posts engaging, informative, and professional.  Use a friendly and approachable tone.  Include relevant hashtags at the 
end of the post.
"""

prompt = f"""
Tell me three news stories about China published on reliable sources outside China in past week, and summarize each.  
Make the summaries engaging, informative, and professional.  Use a friendly and approachable tone.  Translate the news into Chinese.  
Include relevant hashtags at the end of the post. Add links to each story.  Each link should be directly related to the news story rather than a general or generic link.
{search_results}

"""


completion = client.chat.completions.create(
    model="gpt-5.4-2026-03-05",
    messages=[
        {"role": "system", "content": system_message},
        {"role": "user", "content": prompt.format(search_results=search_results)}
    ],
    temperature=0.7,
    max_completion_tokens=700
    )

chat_response=completion.choices[0].message.content
display(Markdown(chat_response))
```
